```python
import pandas as pd
df = pd.DataFrame({
 "nik": ["5171012501990001", "5171014502000002", "5171010703980003"],
 "nama": ["I Wayan Adi", "Ni Kadek Sari", "I Made Bagus"],
 "email": ["adi@example.com", "sari@example.com", "bagus@example.com"],
 "tgl_lahir": ["1999-01-25", "2000-02-05", "1998-03-07"],
 "kode_pos": ["80361", "80361", "80234"],
 "jenis_kelamin": ["L", "P", "L"],
 "rating": [4, 5, 3],
})
klasifikasi = {
 "nik": "PII langsung", "nama": "PII langsung", "email": "PII langsung",
 "tgl_lahir": "Quasi-identifier", "kode_pos": "Quasi-identifier",
 "jenis_kelamin": "Quasi-identifier", "rating": "Non-PII",
}
print(pd.Series(klasifikasi, name="Kategori").to_frame())
```

```python
import hashlib
SALT = "kunci_rahasia_proyek_2026" # simpan di luar repository!
def pseudonim(nilai, salt=SALT, panjang=12):
 return hashlib.sha256((salt + str(nilai)).encode()).hexdigest()[:panjang]
def mask_email(e):
 user, domain = e.split("@")
 return user[0] + "*"*(len(user)-1) + "@" + domain
aman = df.copy()
aman["id_subjek"] = df["nik"].apply(pseudonim)
aman["email"] = df["email"].apply(mask_email)
aman = aman.drop(columns=["nik", "nama"])
print(aman)
print("\nKonsistensi pseudonim:", pseudonim("5171012501990001") == aman.loc[0,
"id_subjek"])
```

```python
aman["tahun_lahir"] = pd.to_datetime(df["tgl_lahir"]).dt.year
aman["kelompok_usia"] = pd.cut(2026 - aman["tahun_lahir"],
 bins=[0, 24, 34, 44, 200],
labels=["<=24", "25-34", "35-44", "45+"])
aman["wilayah"] = df["kode_pos"].str[:3] + "xx"
aman = aman.drop(columns=["tgl_lahir", "tahun_lahir"])
qi = ["kelompok_usia", "wilayah", "jenis_kelamin"]
ukuran_kelas = aman.groupby(qi, observed=True).size()
print(ukuran_kelas)
print("\nk-anonymity =", ukuran_kelas.min(), "(k=1 berarti ada individu yang unik)")
```

```python
from urllib.robotparser import RobotFileParser

def boleh_scrape(url_situs, path="/", user_agent="*"):
    rp = RobotFileParser()
    rp.set_url(url_situs.rstrip("/") + "/robots.txt")
    try:
        rp.read()
    except Exception as e:
        return f"Tidak dapat membaca robots.txt: {e}"
    return rp.can_fetch(user_agent, url_situs.rstrip("/") + path)

for situs in ["https://example.com", "https://www.wikipedia.org"]:
    print(situs, "->", boleh_scrape(situs))
```
